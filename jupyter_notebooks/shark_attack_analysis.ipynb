## Hypotheses

The following hypotheses were developed to examine key patterns in the shark attack dataset. These hypotheses focus on geographic distribution, the relationship between human activity and fatality outcomes, overall fatality proportions, and temporal trends in recorded incidents.

Together, these hypotheses allow the analysis to move beyond descriptive exploration and evaluate whether the observed patterns reflect meaningful relationships within the data.

### Hypothesis 1 — Country Distribution

H0₁: Shark attack frequency is independent of the country in which the incident occurred.

H1₁: Shark attack frequency varies significantly between countries.

Test Method:  
Descriptive geographic analysis using country-level incident frequencies and visualisation.


### Hypothesis 2 — Activity and Fatality Outcome

H0₂: Fatality outcome is independent of the activity being performed during the shark attack incident.

H1₂: Fatality outcome is associated with the activity being performed.

Test Method:  
Chi-square test of independence.


### Hypothesis 3 — Fatal vs Non-Fatal Proportion

H0₃: The proportion of fatal shark attacks is equal to the proportion of non-fatal shark attacks.

H1₃: The proportion of non-fatal shark attacks is greater than the proportion of fatal shark attacks.

Test Method:  
Proportion comparison using descriptive statistics.


### Hypothesis 4 — Temporal Trends

H0₄: The number of recorded shark attack incidents has remained constant over time.

H1₄: The number of recorded shark attack incidents has increased over time.

Test Method:
Trend analysis using descriptive statistics and visual inspection.

# 2. Data Preparation

This section prepares the dataset for further analysis. Initial preparation steps are performed to ensure the dataset is suitable for cleaning and analysis.

These steps include removing variables that are not required for the analysis, preparing column names for easier manipulation, and converting certain variables into more appropriate data types. Preparing the dataset in this way helps ensure consistency and improves the efficiency of subsequent data cleaning and feature engineering steps.

## 2.1 Remove Personally Identifiable Information

The dataset contains a column labelled `name`, which includes the names of individuals involved in shark attack incidents. Since this information is not required for the analysis and represents personally identifiable information, the column is removed to maintain data privacy and ensure the dataset used for analysis is anonymised.

```python
# Remove the 'name' column from the dataset.
# This column contains personally identifiable information (PII), which is not needed for the analysis and should be excluded to help preserve privacy and anonymise the dataset.

df = df.drop(columns=["name"])

# Display the first few rows to confirm that the column has been removed from the DataFrame.

df.head()
```

## 2.2 Standardise Column Names

To improve readability and ensure consistency when referencing variables in code, column names are standardised by converting them to lowercase and removing any unnecessary whitespace.

```python
# Standardise all column names by:
# 1. converting them to lowercase
# 2. removing leading and trailing whitespace

# This makes the dataset easier to work with and reduces the risk of coding errors caused by inconsistent naming conventions.

df.columns = df.columns.str.lower().str.strip()

# Display the updated column names to verify that the formatting has been applied correctly.
df.columns
```

## 2.3 Convert the Year Variable

The `year` variable is currently stored as a floating-point data type. Since years are whole-number values, this variable is converted to numeric format to ensure consistency and support time-based analysis.

```python
# Convert the 'year' column to a numeric data type.
# Any non-numeric values will be converted to NaN using errors='coerce'.

df["year"] = pd.to_numeric(df["year"], errors="coerce")

# Convert the column to pandas nullable integer type (Int64).
# This allows integer values while still supporting missing values.

df["year"] = df["year"].astype("Int64")

# Check the data type to confirm the conversion.

df["year"].dtype
```

## 2.4 Convert the Date Variable

The `date` column is converted from text format to a datetime format. This ensures the variable is suitable for chronological analysis and allows dates to be handled more efficiently in pandas.

```python
# Convert the 'date' column to datetime format.
# Invalid or inconsistent date values will be converted to NaT (Not a Time) using errors='coerce'.

# This allows the date variable to be used for time-based analysis, filtering, and sorting.

df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Display the first few values from the 'date' column to verify that the conversion was successful.

df["date"].head()
```

## 2.5 Verify Dataset Structure

After the initial preparation steps, the dataset structure is reviewed to confirm that the transformations have been applied correctly and that the dataset is ready for the cleaning stage.

```python
# Display a summary of the dataset structure after preparation.
# This includes:
# - the number of rows and columns
# - column names
# - non-null counts
# - data types

# This step is used to confirm that:
# - the 'name' column has been removed
# - the 'date' column is now datetime
# - the 'year' column has been converted to numeric format

df.info()
```

# 3. Data Cleaning and Feature Engineering

This section focuses on cleaning the dataset and preparing variables for analysis. Real-world datasets often contain inconsistencies, missing values, and formatting issues that must be addressed before meaningful analysis can be conducted.

The cleaning process includes handling missing values, standardising categorical variables, and improving the overall quality of the dataset. These steps ensure that the dataset is reliable and suitable for exploratory analysis and visualisation in later stages of the project.

## 3.1 Review Missing Values

Before performing cleaning operations, the dataset is re-examined to identify the extent of missing data across each variable. Understanding where missing values occur helps guide decisions on whether data should be removed, imputed, or retained depending on analytical relevance.

```python
# Recalculate the number of missing values in each column.
# This provides an updated overview of dataset completeness before applying cleaning operations.

df.isnull().sum()
```

The missing value analysis highlights several variables with incomplete data. Some variables such as `type`, `country`, and `year` contain relatively few missing values and can be retained without significant impact on the analysis.

Other variables, including `activity`, `sex`, `area`, and `location`, contain moderate levels of missing data but still provide useful contextual information about shark attack incidents.

However, variables such as `age`, `time`, and `species` contain a substantial number of missing values. While these variables remain potentially valuable for analysis, missing entries will need to be considered when interpreting results.

Also, the `fatal_y_n` variable contains no missing values, making it a reliable variable for analysing the outcomes of shark attack incidents.

## 3.2 Remove Records Without a Year

The `year` variable is critical for analysing temporal trends in shark attacks. Records that do not contain a valid year value cannot contribute to time-based analysis and are therefore removed from the dataset.

```python
# Remove rows where the year value is missing.
# These records cannot be used for time-based analysis.

df = df.dropna(subset=["year"])

# Confirm the dataset size after removal.

df.shape
```

Records with missing values in the `year` column were removed because the year variable is essential for analysing temporal trends in shark attack incidents.

A total of 132 records were removed from the dataset, reducing the dataset from 6,890 to 6,758 observations. Removing these records ensures that all remaining entries contain valid temporal information required for time-based analysis.

## 3.3 Standardise the Sex Variable

The `sex` variable contains inconsistent representations of gender, such as "M", "Male", or lowercase variations. These values are standardised to ensure consistent categorisation during analysis.

```python
# View the unique values in the sex column to understand existing variations.

df["sex"].unique()
```

```python
# Standardise the sex column by converting values to uppercase to ensure consistent formatting.

df["sex"] = df["sex"].str.upper()

# Replace invalid or ambiguous entries with NaN so that only valid categories remain.

df["sex"] = df["sex"].replace(["N", ".", "M X 2", "LLI"], np.nan)

# Check the cleaned values

df["sex"].unique()
```

The `sex` variable contained several inconsistent or invalid entries, including values such as "N", ".", and other irregular text strings. These values likely represent data entry errors or ambiguous records.

To ensure consistency, the variable was standardised by converting text to uppercase and replacing invalid entries with missing values. After cleaning, the column contains only valid categories representing male (`M`) and female (`F`) individuals, with missing values where the information was unavailable.

## 3.4 Clean the Age Variable

The `age` column is currently stored as a text variable and may contain non-numeric values. To support quantitative analysis, the column is converted into a numeric format. Any invalid values are converted to missing values.

```python
# Convert the age column to numeric format.
# Invalid entries will be converted to NaN using errors='coerce'.

df["age"] = pd.to_numeric(df["age"], errors="coerce")

# Verify the updated data type.

df["age"].dtype
```

The `age` column was originally stored as a text variable and contained non-numeric values. To support quantitative analysis, the column was converted to numeric format using `pd.to_numeric()`.

Any invalid or non-numeric values were automatically converted to missing values (`NaN`). After conversion, the variable can now be used for statistical analysis and visualisation of age-related patterns in shark attack incidents.

## 3.5 Inspect Activity Categories

The `activity` variable describes the human activity being performed at the time of the incident. Since this field contains free-text descriptions, it may include a wide variety of values that require standardisation.

```python
# View the most common activity values to understand the types of activities recorded.

df["activity"].value_counts().head(20)
```

```python
# Create a simplified activity category variable. This groups similar activities together to make analysis easier.

df["activity_group"] = df["activity"].str.lower()

# Surfing related activities

df.loc[df["activity_group"].str.contains("surf", na=False), "activity_group"] = "surfing"


# Swimming related activities

df.loc[df["activity_group"].str.contains("swim|bathing|wading|standing|treading", na=False), "activity_group"] = "swimming"


# Fishing related activities

df.loc[df["activity_group"].str.contains("fish", na=False), "activity_group"] = "fishing"


# Diving related activities

df.loc[df["activity_group"].str.contains("div", na=False), "activity_group"] = "diving"


# Watercraft activities

df.loc[df["activity_group"].str.contains("kayak|boat|ski|wind|board", na=False), "activity_group"] = "watercraft"


# Everything else

df.loc[~df["activity_group"].isin(["surfing","swimming","fishing","diving","watercraft"]), "activity_group"] = "other"


# Check the new grouped categories

df["activity_group"].value_counts()
```

```python
# Validate that the new activity group variable has been created correctly by displaying the original and new columns side by side for the first few records.

df[["activity", "activity_group"]].head(10)
```

The `activity` variable contained a large number of unique free-text descriptions describing human activities at the time of the incident. Many of these entries referred to similar activities but were recorded using different wording.

To simplify the analysis and improve interpretability, activities were grouped into broader categories including swimming, surfing, fishing, diving, and watercraft-related activities. Any remaining activities that did not fit into these categories were classified as "other".

This grouping reduces the number of activity categories and allows clearer comparisons when analysing patterns in shark attack incidents.

## 3.6 Inspect Shark Species

The `species` variable identifies the shark species involved in each incident. However, this field contains many variations and descriptive entries. Examining the most common values helps determine whether grouping or simplification may be required.

```python
# Display the most common shark species entries to identify inconsistencies or variations.

df["species"].value_counts().head(20)
```

```python
# Create a lowercase version of the species column to make text matching easier.

df["species_group"] = df["species"].str.lower()


# Identify major species

df.loc[df["species_group"].str.contains("white", na=False), "species_group"] = "white shark"
df.loc[df["species_group"].str.contains("tiger", na=False), "species_group"] = "tiger shark"
df.loc[df["species_group"].str.contains("bull", na=False), "species_group"] = "bull shark"


# All other entries containing the word "shark" are grouped as "other shark"

df.loc[df["species_group"].str.contains("shark", na=False), "species_group"] = "other shark"


# Assign the major species again (so they are not overwritten)

df.loc[df["species"].str.contains("White", na=False), "species_group"] = "white shark"
df.loc[df["species"].str.contains("Tiger", na=False), "species_group"] = "tiger shark"
df.loc[df["species"].str.contains("Bull", na=False), "species_group"] = "bull shark"


# Replace missing or uncertain values

df.loc[df["species_group"].isna(), "species_group"] = "unknown"


# Check the grouped results

df["species_group"].value_counts()
```

```python
# Replace invalid or uncertain species labels with "unknown"

df["species_group"] = df["species_group"].replace([
    "invalid",
    "questionable incident",
    "questionable",
    "unidentified species",
    "unidentified"
], "unknown")


# Check the cleaned results again

df["species_group"].value_counts()
```

```python
# Define the allowed species categories

major_species = ["white shark", "tiger shark", "bull shark", "other shark", "unknown"]


# Any value not in the major species list will be grouped as "other shark"

df.loc[~df["species_group"].isin(major_species), "species_group"] = "other shark"


# Check the final result

df["species_group"].value_counts()
```

The `species` column contained numerous inconsistent entries including size descriptions, uncertain identifications, and free-text incident notes. To improve interpretability, the variable was simplified by grouping shark species into broader categories.

Major species such as white sharks, tiger sharks, and bull sharks were retained as separate categories due to their frequency in the dataset. All other shark-related entries were grouped under "other shark", while records without confirmed shark identification were classified as "unknown".

As a result, this approach reduces the complexity of the species variable and allows clearer comparisons in subsequent analysis and visualisations.

Other values are treated as missing data and excluded from fatality analysis.

## 4.2 Shark Attack Incidents Over Time

# 5. Hypothesis Testing

This section applies statistical methods to test whether observed patterns in the dataset are likely to reflect meaningful relationships rather than random variation.

The analysis focuses on hypotheses relating to activity type, fatality outcome, and country distribution. Chi-square tests of independence are used to examine whether relationships exist between categorical variables.

## 5.1 Chi-Square Test: Activity Type and Fatality Outcome

To examine whether fatality outcome is associated with the type of human activity being performed, a chi-square test of independence is applied to the cross-tabulated activity and fatality variables.

The chi-square test is appropriate here because both variables are categorical. The test evaluates whether the observed distribution of fatal and non-fatal incidents across activity categories differs significantly from what would be expected if the variables were independent.

```python
# Import the chi-square test function from scipy.stats.
# This function will be used to test whether there is a statistically significant association between activity type and fatality outcome.

from scipy.stats import chi2_contingency
```

```python
# Create a contingency table using the cleaned activity_group and fatal_y_n variables. This table shows the observed frequencies of fatal and non-fatal incidents across each activity category.

activity_fatal_table = pd.crosstab(df["activity_group"], df["fatal_y_n"])


# Run the chi-square test of independence.
# The function returns:
# - chi2: the chi-square test statistic
# - p: the p-value
# - dof: degrees of freedom
# - expected: the expected frequencies if the variables were independent

chi2, p, dof, expected = chi2_contingency(activity_fatal_table)


# Display the main test results.

print("Chi-square statistic:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)
```

The chi-square test of independence was conducted to determine whether fatality outcome is associated with the type of human activity being performed during a shark attack incident.

The chi-square test produced a statistically significant result (χ²(5) = 452.96, p < 0.001).

Because the p-value is far below the commonly used significance threshold of 0.05, the null hypothesis is rejected. This indicates that fatality outcome is not independent of activity type, and that there is a statistically significant association between the activity being performed and the likelihood of a fatal outcome.

This result supports the earlier exploratory analysis, which suggested that certain activities, such as swimming and watercraft use, have higher proportions of fatal incidents compared with activities such as surfing. The findings suggest that the type of activity performed in the water may influence the severity of shark attack incidents.

```python
# Convert the expected frequencies into a DataFrame so they can be compared with the observed contingency table.

expected_df = pd.DataFrame(
    expected,
    index=activity_fatal_table.index,
    columns=activity_fatal_table.columns
)

expected_df
```

The expected frequency table shows the counts that would be expected in each category if activity type and fatality outcome were independent.

Comparing the expected frequencies with the observed contingency table reveals substantial differences across several activity categories. For example, the number of fatal incidents associated with surfing is considerably lower than expected under independence, while activities such as swimming and watercraft show higher numbers of fatal incidents than expected.

These deviations between the observed and expected frequencies contribute to the large chi-square statistic obtained in the test results. Together with the extremely small p-value, this provides strong evidence that fatality outcome is associated with the type of human activity being performed during shark attack incidents.

# 6. Ethical Considerations

### Data Privacy and Anonymisation

The original dataset included personally identifiable information, such as the names of individuals involved in shark attack incidents. In order to protect individual privacy and minimise the unnecessary processing of personal data, the `name` variable was removed during the data cleaning stage.

This decision aligns with the principles of data minimisation and privacy protection established under the General Data Protection Regulation (GDPR) (European Union, 2016). GDPR emphasises that personal data should only be processed when it is necessary for a specific analytical purpose. In this project, the identity of individuals involved in incidents is not required for analysing patterns in shark attack occurrences. Therefore, removing personal identifiers ensures that the analysis focuses on aggregated patterns rather than identifiable individuals.

This approach is also consistent with guidance from the Information Commissioner’s Office (ICO, 2023), which recommends limiting the use of personal data when it is not essential for the intended analysis. By anonymising the dataset, the project reduces the risk of unnecessary exposure of personal information while maintaining the integrity of the analytical process.

### Responsible Interpretation of Wildlife Data

Shark attack data can influence public perceptions of sharks and marine ecosystems. Media coverage of shark incidents often portrays sharks as dangerous predators, which can contribute to exaggerated perceptions of risk among the public (Neff & Hueter, 2013). However, sharks play an essential ecological role as apex predators that help regulate marine food chains and maintain ecosystem stability (Dulvy et al., 2014).

For this reason, the analysis presented in this project aims to interpret shark attack data responsibly. The findings emphasise that shark attacks are relatively rare events compared with the large number of people who participate in ocean-based recreational activities each year. Providing this context helps avoid reinforcing negative stereotypes about sharks and encourages a more balanced understanding of human–wildlife interactions.

### Data Limitations and Reporting Bias

The dataset used in this analysis consists of historical records of reported shark attack incidents. As with many historical datasets, the data may contain reporting biases. Incidents occurring in countries with stronger media coverage or more developed reporting systems may be more likely to be documented than incidents occurring in regions with limited documentation infrastructure.

As a result, the geographic distribution of shark attack incidents observed in the dataset may partly reflect differences in reporting practices rather than purely differences in the frequency of shark–human encounters. Recognising these limitations is important to ensure that conclusions drawn from the data are interpreted cautiously.

### Ethical Communication of Findings

The findings of this analysis are intended to be communicated through an interactive dashboard. When presenting analytical results, it is important that visualisations accurately represent the underlying data and do not exaggerate risks associated with sharks or marine environments.

Clear visual communication, appropriate context, and acknowledgement of dataset limitations help ensure that the insights derived from this project are interpreted responsibly. Ethical data communication is essential for preventing misinterpretation and ensuring that analytical findings contribute positively to public understanding of shark–human interactions.

# 9. Discussion and Conclusions

This section summarises the key findings of the analysis and reflects on the implications of the results. It also discusses the limitations of the dataset and analytical approach used in this project before presenting final conclusions regarding patterns in shark–human interactions.

## 9.1 Discussion

The analysis conducted in this project aimed to explore patterns in global shark attack incidents using both statistical methods and interactive visualisation techniques. The results highlight several notable trends within the dataset.

First, the temporal analysis revealed that the number of recorded shark attack incidents has generally increased over time. However, previous research suggests that this increase is likely influenced by factors such as population growth in coastal regions, increased participation in ocean-based recreational activities, and improved reporting of shark incidents rather than an actual increase in shark aggression (Ferretti et al., 2015).

Second, the geographic distribution of incidents indicates that shark attacks are concentrated in coastal regions with high levels of ocean recreation, particularly in countries such as the United States, Australia, and South Africa. These regions combine favourable shark habitats with dense coastal populations and high levels of water-based recreational activities, which increases the likelihood of shark–human interactions (Neff & Hueter, 2013).

The activity-based analysis further demonstrates that swimming and surfing account for a large proportion of incidents. These activities typically occur near the ocean surface, where sharks may mistake humans for prey or become attracted to unusual movement patterns in the water (Chapman & McPhee, 2016).

In addition, the analysis of fatal versus non-fatal outcomes revealed that most shark attacks are non-fatal. Research suggests that many shark encounters involve exploratory bites rather than deliberate predatory behaviour, which may explain why fatal outcomes represent a smaller proportion of incidents (Neff & Hueter, 2013).

Finally, the statistical hypothesis testing conducted earlier in the project identified a significant relationship between activity type and fatality outcome. This suggests that certain activities may influence the severity of shark encounters, potentially due to differences in environmental conditions, distance from shore, or duration of exposure in the water.

Together, these findings provide a clearer understanding of the conditions under which shark–human interactions are more likely to occur and the factors that may influence the severity of such incidents.

## 9.2 Limitations

Despite the insights gained from this analysis, several limitations should be acknowledged.

Predictive modelling was not included in this project because the available dataset primarily contains historical incident records and categorical variables, which are better suited to exploratory analysis and statistical testing rather than predictive forecasting.

One key limitation relates to the completeness and accuracy of the dataset. Many records contain missing or incomplete information, particularly with regard to shark species identification, victim age, and specific activity details. A substantial proportion of incidents are classified with an "Unknown" species, which limits the ability to accurately assess species-specific risk patterns.

Additionally, the dataset relies on historical records that may be subject to reporting bias. Earlier incidents may have been underreported due to limited documentation, media coverage, or scientific recording practices. As a result, the apparent increase in incidents over time may partially reflect improved reporting rather than a true rise in shark attack frequency.

Another limitation is the lack of environmental variables within the dataset. Factors such as water temperature, ocean currents, prey availability, and seasonal patterns can influence shark behaviour, but these variables were not available for inclusion in the analysis.

Finally, the statistical analysis focused on relationships within the recorded data and does not establish causal relationships between variables. While certain activities appear associated with higher fatality proportions, further research would be required to fully understand the mechanisms behind these patterns.

## 9.3 Future Work

While this project provides valuable insights into patterns of shark–human interactions, several opportunities exist for extending the analysis in future work.

First, additional environmental variables could be incorporated into the dataset. Factors such as water temperature, ocean currents, seasonal migration patterns, and prey availability may influence shark behaviour and could improve the understanding of conditions under which shark encounters occur.

Second, future studies could integrate spatial analysis techniques using geographic information systems (GIS) to explore regional patterns of shark activity in greater detail. This could help identify environmental hotspots where shark–human interactions are more likely.

Third, with access to richer environmental and behavioural datasets, predictive modelling techniques could be explored to estimate the likelihood of shark encounters under certain conditions. However, such models would require more detailed and structured data than the historical incident records used in this project.

Finally, expanding the dataset to include more recent and consistently reported incident records could improve the reliability of future analyses and support more advanced statistical modelling.

Overall, these extensions could provide deeper insights into shark–human interactions and support the development of evidence-based strategies for improving ocean safety and shark conservation.

## 9.4 Conclusion

This project explored global shark attack patterns using a combination of data cleaning, exploratory data analysis, statistical hypothesis testing, and interactive dashboard visualisation.

The findings indicate that shark attack incidents have increased in recorded frequency over time, likely reflecting increased human activity in marine environments and improvements in incident reporting. Geographic analysis highlights the concentration of incidents in coastal regions with high levels of ocean recreation, while activity-based analysis demonstrates that swimming and surfing are commonly associated with shark encounters.

The analysis also shows that most shark attacks are non-fatal, with fatal incidents representing a smaller proportion of total cases. Statistical testing further identified a relationship between activity type and fatality outcomes, suggesting that certain human behaviours may influence the severity of encounters.

The interactive Power BI dashboard developed in this project provides an accessible tool for exploring these patterns and enables users to dynamically examine shark attack incidents across multiple dimensions.

Overall, this project contributes to a better understanding of shark–human interactions and highlights the importance of data-driven analysis in interpreting patterns within environmental and wildlife-related datasets.The project also demonstrates how data analytics techniques, including data cleaning, statistical testing, and interactive visualisation, can be applied to real-world environmental datasets to support evidence-based understanding of complex human–wildlife interactions.
